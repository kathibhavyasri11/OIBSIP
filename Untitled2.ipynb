```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("retail_sales.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df = df.dropna()

df = df.drop_duplicates()

print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df.head()
```

```python
df.describe()
```

```python
average_purchase_value = df["Total Amount"].mean()

print("Average Purchase Value:", average_purchase_value)
```

```python
purchase_frequency = df.groupby("Customer ID").size()

print(purchase_frequency.head())
```

```python
customer_lifetime_value = df.groupby("Customer ID")["Total Amount"].sum()

print(customer_lifetime_value.head())
```

```python
reference_date = df["Date"].max() + pd.Timedelta(days=1)

print(reference_date)
```

```python
rfm = df.groupby("Customer ID").agg(
    Recency=("Date", lambda x: (reference_date - x.max()).days),
    Frequency=("Transaction ID", "count"),
    Monetary=("Total Amount", "sum")
)

rfm.head()
```

```python
rfm.describe()
```

```python
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm[["Recency", "Frequency", "Monetary"]]
)
```

```python
print(rfm_scaled[:5])
```

```python
inertia = []

for k in range(1, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_scaled)

    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertia,
    marker="o"
)

plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.grid()

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans.fit_predict(rfm_scaled)

rfm.head()
```

```python
cluster_profile = rfm.groupby("Cluster").mean()

print(cluster_profile)
```

```python
customer_count = rfm.groupby("Cluster").size()

print(customer_count)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rfm,
    x="Frequency",
    y="Monetary",
    hue="Cluster",
    palette="viridis",
    s=100
)

plt.title("Customer Clusters: Frequency vs Monetary")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rfm,
    x="Recency",
    y="Monetary",
    hue="Cluster",
    palette="viridis",
    s=100
)

plt.title("Customer Clusters: Recency vs Monetary")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=rfm,
    x="Cluster",
    hue="Cluster",
    legend=False
)

plt.title("Number of Customers per Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")

plt.show()
```

```python
cluster_names = {
    0: "Regular Customers",
    1: "High-Value Customers",
    2: "At-Risk Customers"
}

rfm["Customer Segment"] = rfm["Cluster"].map(cluster_names)

rfm.head()
```

```python
segment_profile = rfm.groupby("Customer Segment").agg(
    Average_Recency=("Recency", "mean"),
    Average_Frequency=("Frequency", "mean"),
    Average_Monetary=("Monetary", "mean"),
    Number_of_Customers=("Customer Segment", "count")
)

segment_profile
```
