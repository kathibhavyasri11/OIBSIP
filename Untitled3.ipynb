```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("retail_sales.csv")

# Display first 5 rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Null Count": df.isnull().sum().values,
    "Null Percentage": (df.isnull().mean() * 100).round(2).values,
    "Unique Values": df.nunique().values
})

quality_report
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df[df.duplicated()]
```

```python
missing_data = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing_data = missing_data[missing_data["Missing Values"] > 0]

missing_data
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

duplicates_removed = before_duplicates - after_duplicates

print("Duplicates removed:", duplicates_removed)
print("Rows remaining:", len(df))
```

```python
if "Gender" in df.columns:
    df["Gender"] = (
        df["Gender"]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace({
            "m": "Male",
            "male": "Male",
            "man": "Male",
            "f": "Female",
            "female": "Female",
            "woman": "Female"
        })
    )

df["Gender"].value_counts()
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.title()
```

```python
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

df["Date"].head()
```

```python
print(df["Date"].dtype)
```

```python
numeric_columns = ["Salary", "Revenue", "Price"]

for col in numeric_columns:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("$", "", regex=False)
            .str.strip()
        )

        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
print(df.dtypes)
```

```python
id_columns = ["ID", "Employee_ID", "Customer_ID"]

for col in id_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_summary)

outlier_report
```

```python
from scipy.stats import zscore

numeric_columns = df.select_dtypes(include=np.number).columns

zscore_report = []

for col in numeric_columns:
    z_scores = zscore(df[col], nan_policy="omit")

    outlier_count = np.sum(np.abs(z_scores) > 3)

    zscore_report.append({
        "Column": col,
        "Outliers (Z > 3)": outlier_count
    })

zscore_report = pd.DataFrame(zscore_report)

zscore_report
```

```python
print("Final Data Types:\n")
print(df.dtypes)
```

```python
original_df = pd.read_csv("retail_sales.csv")
```

```python
before_rows = len(original_df)
after_rows = len(df)

before_duplicates = original_df.duplicated().sum()
after_duplicates = df.duplicated().sum()

before_nulls = original_df.isnull().sum().sum()
after_nulls = df.isnull().sum().sum()

before_columns = original_df.shape[1]
after_columns = df.shape[1]

summary = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Column Count",
        "Null Count",
        "Duplicate Count"
    ],
    "Before Cleaning": [
        before_rows,
        before_columns,
        before_nulls,
        before_duplicates
    ],
    "After Cleaning": [
        after_rows,
        after_columns,
        after_nulls,
        after_duplicates
    ]
})

summary
```

```python
dtype_comparison = pd.DataFrame({
    "Column": df.columns,
    "Original Type": [
        original_df[col].dtype if col in original_df.columns else "New"
        for col in df.columns
    ],
    "Cleaned Type": [
        df[col].dtype
        for col in df.columns
    ]
})

dtype_comparison
```

```python
print("========== FINAL DATA QUALITY CHECK ==========")

print("\nShape:", df.shape)

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nTotal Duplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values by Column:")
print(df.isnull().sum())
```

```python
output_file = "cleaned_dataset.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully as:", output_file)
```
